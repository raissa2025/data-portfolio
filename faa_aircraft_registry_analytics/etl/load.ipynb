```python
import pandas as pd
from sqlalchemy import create_engine
from dotenv import load_dotenv
import os


df_master=pd.read_csv(aircraft_master_clean.csv)
df_ref=pd.read_csv(aircraft_reference_clean.csv)
df_analytics=pd.read_csv(faa_analytics_dataset.csv)

# Load environment variables
load_dotenv()

DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")
DB_USER = os.getenv("DB_USER")
DB_PASSWORD = os.getenv("DB_PASSWORD")

# Create SQLAlchemy engine
engine = create_engine(
    
     "postgresql+psycopg2://postgres:postgres@localhost:5432/faa_analytics"
)

# Load tables
df_master.to_sql(
    "aircraft_master",
    engine,
    if_exists="replace",
    index=False
)

df_ref.to_sql(
    "aircraft_reference",
    engine,
    if_exists="replace",
    index=False
)

df_analytics.to_sql(
    "faa_analytics",
    engine,
    if_exists="replace",
    index=False
)

print(" Data successfully loaded into PostgreSQL")
```
