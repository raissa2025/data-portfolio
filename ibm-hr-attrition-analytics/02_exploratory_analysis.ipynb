# IBM HR Analytics: Employee Attrition – Exploratory Data Analysis (EDA)

# This notebook explores key HR factors that influence employee attrition.

# 1. Import Libraries

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sns.set(style="whitegrid")
```

# 2. Load Cleaned Dataset

```python
df=pd.read_csv("data\hr_clean.csv")
```

# 3. Exploratory Data Analysis (EDA)

# 3.1 Attrition by Department

```python
dept_attrition = df.groupby('department')['attrition_flag'].mean()
dept_attrition
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x=dept_attrition.index, y=dept_attrition.values, hue=dept_attrition.index, palette="Set1")
plt.title("Attrition Rate by Department")
plt.ylabel("Attrition Rate")
plt.xlabel("Department")
plt.show()
```

# 3.2 Attrition by Job Role

```python
role_attrition = df.groupby('job_role')['attrition_flag'].mean().sort_values(ascending=False)
role_attrition
```

```python
plt.figure(figsize=(16,7))
sns.barplot(x=role_attrition.index, y=role_attrition.values, hue=role_attrition.index, palette="Set1")
plt.xticks(rotation=45)
plt.title("Attrition Rate by Job Role")
plt.ylabel("Attrition Rate")
plt.xlabel("Job Role")
plt.tight_layout()
plt.show()
```

# 3.3 Overtime Impact on Attrition

```python
overtime_attrition = df.groupby('overtime')['attrition_flag'].mean()
overtime_attrition
```

```python
plt.figure(figsize=(6,4))
sns.barplot(x=overtime_attrition.index, y=overtime_attrition.values, hue= overtime_attrition.index, palette="Set1")
plt.title("Attrition Rate by Overtime")
plt.ylabel("Attrition Rate")
plt.xlabel("Overtime")
plt.show()
```

# 3.4 Attrition by Job Satisfaction

```python
satisfaction_attrition = df.groupby('job_satisfaction')['attrition_flag'].mean()
satisfaction_attrition
```

```python
plt.figure(figsize=(7,5))
sns.barplot(x=satisfaction_attrition.index, y=satisfaction_attrition.values,hue= satisfaction_attrition.index, palette="Set1")
plt.title("Attrition Rate by Job Satisfaction")
plt.ylabel("Attrition Rate")
plt.xlabel("Job Satisfaction Level")
plt.show()
```

# 3.5 Correlation with Attrition

```python
corr_attrition = df.corr(numeric_only=True)['attrition_flag'].sort_values(ascending=False)
corr_attrition
```
