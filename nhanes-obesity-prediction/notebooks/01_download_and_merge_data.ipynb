# NHANES Data Analysis & Obesity Prediction
## 01 — Download and Merge Data

## This notebook prepares the foundational dataset used throughout the project. It downloads NHANES demographic (DEMO) and body‑measurement (BMX) files from the 2017–2020 and 2021–2023 cycles, selects the relevant variables, and merges them into a single, analysis‑ready dataset. The output of this notebook is `merged_data.csv`, which serves as the starting point for all subsequent cleaning, exploration, and modeling steps.

## 1. Import Libraries 

```python
import pandas as pd
```

## 2. Load NHANES XPT Files

```python
# Read the .xpt files
bmx_17_20 = pd.read_sas("data\P_BMX-17-20.xpt")
demo_17_20 = pd.read_sas("data\P_DEMO-17-20.xpt")
bmx_21_23 = pd.read_sas("data\BMX_L21-23.xpt")
demo_21_23 = pd.read_sas("data\DEMO_L21-23.xpt")
```

## 3. Select Relevant Variables

```python
# Variables to keep
demo_vars = ["SEQN", "RIAGENDR", "RIDAGEYR", "RIDRETH3", "DMDEDUC2", "DMDMARTZ", "INDFMPIR"]
bmx_vars  = ["SEQN", "BMXBMI", "BMXWAIST", "BMXARMC", "BMXHIP", "BMXHT"]

# Apply selections
demo_17_20 = demo_17_20[demo_vars]
demo_21_23 = demo_21_23[demo_vars]

bmx_17_20 = bmx_17_20[bmx_vars]
bmx_21_23 = bmx_21_23[bmx_vars]
```

## 4. Stack Cycles Together

```python
# Combine cycles
demo_all = pd.concat([demo_17_20, demo_21_23], ignore_index=True)
bmx_all  = pd.concat([bmx_17_20, bmx_21_23], ignore_index=True)
```

## 5. Merge DEMO + BMX

```python
merged = pd.merge(demo_all, bmx_all, on="SEQN", how="inner")

# Inspect
print(merged.head())
print(merged.shape)
```

## 6. Save Final Dataset

```python
merged.to_csv("merged_data.csv", index=False)
```
