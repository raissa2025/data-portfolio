## 03 — EDA Part 1: Descriptive Statistics

## This notebook summarizes the key demographic and clinical variables in the cleaned NHANES dataset. Both continuous and categorical variables are explored to understand the distribution of the adult population included in the study.

## 1. Load the Cleaned Dataset

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("\data\clean_merged_data.csv")
```

## 2. Descriptive Statistics for Continuous Variables

```python
# Describe continuous variables
num_vars = ["RIDAGEYR", "INDFMPIR", "BMXWAIST", "BMXARMC", "BMXHIP", "BMXHT"]
desc_stat = df[num_vars].describe().T
print(desc_stat)
desc_stat.to_csv("descriptive_stats.csv")
```

## Interpretation

###  The average age (RIDAGEYR) is around 50 years, with a wide range from 18 to 80, showing a diverse adult population.

###  The mean income ratio (INDFMPIR) is 2.64, with values ranging from very low to a maximum of 5, indicating socioeconomic diversity.

### - Body measurements (waist, arm, hip, height):  
###  These variables show expected variation in an adult population.

###  - Waist circumference averages 100 cm

###  - Hip circumference averages 107 cm

###  - Height averages 167 cm

### These values align with typical U.S. adult anthropometric patterns.

## 3. Descriptive Statistics for Categorical Variables

```python
# Describe categorical variables
mappings = {
 "RIAGENDR": {1.0: "Male", 2.0: "Female"},
 "DMDMARTZ": {1.0: "Married", 2.0: "Widowed", 3.0: "Divorced/Separated", 77.0: "Refused", 99.0: "Don't know"},

 "RIDRETH3": {1.0: "Mexican American", 2.0: "Other Hispanic", 3.0: "Non-Hispanic White", 4.0: "Non-Hispanic Black",
    6.0: "Non-Hispanic Asian", 7.0: "Other/Multi-Race"},

 "DMDEDUC2": {1.0: "Less than 9th grade", 2.0: "9–11th grade", 3.0: "High school/GED", 4.0: "Some college/AA",
    5.0: "College graduate or above", 7.0: "Refused", 9.0: "Don't know"},
 
 "Obese": {0.0: "Not Obese", 1.0: "Obese"}}

df_labeled = df.copy()
for col, mapping in mappings.items():
    df_labeled[col] = df_labeled[col].map(mapping)

cat_vars = list(mappings.keys())

def count_percent(series):
    counts = series.value_counts(dropna=False)
    percentages = series.value_counts(normalize=True, dropna=False) * 100
    return pd.DataFrame({"Count": counts, "Percentage": percentages.round(2)})

cat_summary = {col: count_percent(df_labeled[col]) for col in cat_vars}
all_summary = pd.concat(cat_summary, axis=0)
print(all_summary)
```

## Interpretation
###  The sample is slightly female‑dominant.
###  Most adults are married, with a notable proportion widowed or divorced.
###  The dataset reflects a diverse population, with large Non‑Hispanic White, Black, and Hispanic groups.
###  Most participants completed high school or some college, with a smaller share holding a college degree.
###  About two‑fifths of adults meet the BMI criteria for obesity.
